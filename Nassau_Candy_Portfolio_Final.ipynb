The original date strings include values such as `21-11-2024`, which confirms the day-month-year format. Parse the Order Date once for monthly sales and profit trends. Keep Ship Date for the separate quality investigation below.

## 6. Shipping Date Quality Investigation

This is a **data-quality check**, not a delivery-performance comparison. We calculate duration once, review its distribution, check invalid/negative values, and compare order years with ship years. Do not use the duration results as operational KPIs unless the source data is verified.

```python
# Parse the raw Ship Date values and calculate duration once
# Both date columns in the supplied raw examples use DD-MM-YYYY.
df['Ship Date Parsed'] = pd.to_datetime(
    df['Ship Date'].astype(str).str.strip(),
    format='%d-%m-%Y',
    errors='coerce'
)

df['Shipping Duration (Days)'] = (
    df['Ship Date Parsed'] - df['Order Date']
).dt.days

print("Invalid Order Date values:", df['Order Date'].isna().sum())
print("Invalid Ship Date values:", df['Ship Date Parsed'].isna().sum())
print("Negative durations:", (df['Shipping Duration (Days)'] < 0).sum())
print("Zero durations:", (df['Shipping Duration (Days)'] == 0).sum())
print("Durations greater than 90 days:", (df['Shipping Duration (Days)'] > 90).sum())
print("Total records:", len(df))

print("\nShipping duration summary (days):")
display(df['Shipping Duration (Days)'].describe())

print("\nOrder Date range:", df['Order Date'].min(), "to", df['Order Date'].max())
print("Ship Date range:", df['Ship Date Parsed'].min(), "to", df['Ship Date Parsed'].max())

print("\nOrder Year vs Ship Year:")
display(pd.crosstab(df['Order Date'].dt.year, df['Ship Date Parsed'].dt.year))

print("\nSample rows for review:")
display(df[['Order Date', 'Ship Date', 'Ship Date Parsed', 'Shipping Duration (Days)', 'Ship Mode']].head(10))
```

### Shipping Investigation – Findings & Limitations

Use the results produced above to verify these statements after running the notebook:

- Check whether the ship dates fall substantially later than the order dates.
- If all or nearly all shipping durations exceed 90 days, treat this as an anomaly requiring source validation.
- Do not infer delivery speed or rank shipping modes by average duration until the date fields are verified.
- Do not manually change dates or fabricate replacement dates.

**Limitation:** The shipping date fields may be inconsistent with expected delivery timing. This analysis documents the issue; it does not establish the cause or correct the source records.
